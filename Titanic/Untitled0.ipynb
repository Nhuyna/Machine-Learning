```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python

train_pd = pd.read_csv('/content/drive/MyDrive/Machine_learning/Titanic/train.csv')
test_pd = pd.read_csv('/content/drive/MyDrive/Machine_learning/Titanic/test.csv')

display(train_pd)
```

```python
survised = train_pd['Survived']
train_pd= train_pd.drop(['Survived'], axis=1)

display(train_pd)
```

```python
train_idx=train_pd['PassengerId']
test_idx=test_pd['PassengerId']

combined_pd=pd.concat([train_pd,test_pd]).reset_index(drop=True)
display(combined_pd)
```

```python
print('Null value')
print(combined_pd.isnull().sum())
```

```python
y= combined_pd['Age'].value_counts()
x=y.index.values

plt.figure(figsize=(10,5))
plt.bar(x,y)
plt.title("All Ages")
plt.show()
```

```python
class_age=combined_pd[['Pclass','Age']]

for i in range(1,4):
  plt.figure(figsize=(10,5))
  plt.title('Class' + str(i))
  y=class_age[class_age['Pclass']==1]['Age'].value_counts()
  x=y.index.values
  plt.bar(x,y)
  plt.show()

medians= class_age.groupby('Pclass').median()
print(medians)
```

```python
for i in range(3):
    # Tìm chỉ số (index) các hàng có Pclass = i+1 và Age bị thiếu
    idx = np.where((combined_pd['Pclass'] == i+1) & (combined_pd['Age'].isnull()))[0]

    # Gán giá trị median tương ứng
    combined_pd.loc[idx, 'Age'] = medians.values[i][0]

display(combined_pd)
```

```python
display(combined_pd[combined_pd['Fare'].isnull()])

sin_fares= combined_pd[combined_pd['Pclass']]
print('Median: ', sin_fares.median())
```
